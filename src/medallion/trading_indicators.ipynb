```python
from pyspark.sql import SparkSession, DataFrame
from trading import sma, ema, rsi, macd
```

```python
spark = SparkSession.builder \
    .appName("LocalTest") \
    .master("local[*]") \
    .getOrCreate()
```

```python
base_path: str = "../../DOGEUSD_silver"
df: DataFrame = spark.read.parquet(base_path)
```

```python
df = sma(df, n=200)
df = ema(df, n=50)
df = rsi(df, n=14)
df = macd(df, fast=12, slow=26, signal=9)
```
