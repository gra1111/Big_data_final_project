```python
from trading_view.trading_view import *
import os
```

```python
tv = TradingViewData()
```

```python
data = tv.get_hist(
    "DOGEUSD",
    "BINANCE",
    interval=Interval.daily,
    n_bars=365*4+56,
    extended_session=False,
)
```

```python
data
```

```python
initial = datetime.datetime(2022, 1, 1, 1, 0, 0)
for year in range(2022, 2026):
    os.makedirs(f"../../data/{year}", exist_ok=True)
    for month in range(1, 13):
        init_date = datetime.datetime(year, month, 1, 1, 0, 0)
        if month == 12:
            end_date = datetime.datetime(year+1, 1, 1, 1, 0, 0)
        else:
            end_date = datetime.datetime(year, month+1, 1, 1, 0, 0)
        os.makedirs(f"../../DOGEUSD/year={year}/month={month}", exist_ok=True)
        data.loc[init_date:end_date].iloc[:-1, 1:].to_csv(f"../../DOGEUSD/year={year}/month={month}/dogeusd_{year}_{month}.csv")
```
