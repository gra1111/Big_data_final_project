```python
import boto3
```

```python
profile = "Comillas-BIGDATA-Alumnos-443370699579"
session = boto3.Session(profile_name=profile)
client = session.client('glue')
```

```python
name = "DodgyCrawler"
role = "DodgyGlueCrawler"
dbname = "trade_data_dodgyconsulting"

path = "s3://dodgy-consulting-trading-bucket/DOGEUSD"

try:
    response = client.create_crawler(
        Name=name,
        Role=role,
        DatabaseName=dbname,
        Description="Dodgy Crawler for Glue",
        Targets={
            "S3Targets":[
                {
                    "Path": path,
                },
            ],
        },
    )
    
    print("SUCCESS - STARTING CRAWLER...")
    response = client.start_crawler(Name="DodgyCrawler")
except:
    print("ERROR - DELETING CRAWLER")
    client.delete_crawler(Name="DodgyCrawler")
```
