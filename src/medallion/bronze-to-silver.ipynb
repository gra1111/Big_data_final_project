```python
from pyspark.sql import SparkSession
```

```python
sc = SparkSession.builder.appName("localTest").master("local[*]").getOrCreate()
```

```python
df = sc.read.option("header", True).option("inferSchema", True).csv("../../DOGEUSD")
```

```python
df.printSchema()
```

```python
df.show()
```

```python

df.write.partitionBy("year", "month").mode("overwrite").format("parquet").save("../../DOGEUSD_silver")
```

```python
df2 = sc.read.parquet("../../DOGEUSD_silver")
```

```python
df2.show()
```

```python
df2.printSchema()
```
